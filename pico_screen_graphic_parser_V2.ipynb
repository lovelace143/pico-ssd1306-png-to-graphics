```python
from PIL import Image

#blegh
filename = 'handOff.png'
xDimension = 128
yDimension = 32
#^^ default dimensions, default file

img = Image.open(filename)
img.thumbnail((xDimension, yDimension), Image.Resampling.NEAREST)
img = img.convert("1")
myWidth = img.width
pixels = img.load()
stringout = ''
stringout_w = 'B' #a black-dominant image's string
stringout_b = 'W' #a white-dominant image's string
xsize = 1
largestX_W = -1
largestX_B = -1
width, height = img.size

img_pix_sum = 0

for y in range(height):
    largestX_W = -1
    largestX_B = -1
    first_W = False
    first_B = False
    for x in range(width): #read each pixel group
        if(x > largestX_W and pixels[x, y] == 255):
            img_pix_sum += 1
            xsize = 1
            seenBlack = False
            while(x+xsize < width and seenBlack == False):
                if(pixels[x+xsize, y] == 255):
                    img_pix_sum += 1
                    xsize += 1
                    largestX_W = x+xsize
                else:
                    seenBlack = True
            if(not first_W):
              first_W = True
              stringout_w += f"{x},{xsize}"
            else:
              stringout_w += f",{x},{xsize}"
        elif(x > largestX_B and pixels[x, y] == 0):
            xsize = 1
            seenWhite = False
            while(x+xsize < width and seenWhite == False):
                if(pixels[x+xsize, y] == 0):
                    xsize += 1
                    largestX_B = x+xsize
                else:
                    seenWhite = True
            if(not first_B):
              first_B = True
              stringout_b += f"{x},{xsize}"
            else:
              stringout_b += f",{x},{xsize}"
    if(y != height-1):
      stringout_b += 'n'
      stringout_w += 'n'

if(img_pix_sum / (width * height) >= 0.5):
  stringout += stringout_b
  #print(f"{len(stringout_b)} is shorter than {len(stringout_w)}")
else:
  stringout += stringout_w
  #print(f"{len(stringout_w)} is shorter than {len(stringout_b)}")

with open(filename+".txt", "w", encoding="utf-8") as file:
    file.write(stringout)
```

```python
#code example of how a single frame should be read in the PICO under micropython:

print("display = ssd1306.SSD1306_SPI(...)")
noBW = stringout[1:]
BWChar = stringout[0]
print(noBW)
if BWChar == "B":
  print("display.fill(0)")
else:
  print("display.fill(1)")
lines = noBW.split("n")
yPos = 0
for line in range(len(lines)):
  lineArr = lines[line].split(",")
  for i in range(int(len(lineArr)/2)): #automatically skips empty lines which get rounded to 0
    xPos = int(lineArr[(i*2)])
    xSize = int(lineArr[(i*2)+1])
    print(f"display.fill_rect({xPos}, {yPos}, {xSize}, 1, 1)")
  yPos += 1
```

```python
#animation frames in a zip file
import zipfile
from PIL import Image
animZip = zipfile.ZipFile('frames.zip') #default
stringout = ''
for filename in animZip.namelist():
  img = Image.open(animZip.open(filename))
  img.thumbnail((xDimension, yDimension), Image.Resampling.NEAREST)
  img = img.convert("1")
  pixels = img.load()
  stringout_w = 'FB' #a black-dominant image's string
  stringout_b = 'FW' #a white-dominant image's string
  xsize = 1
  largestX_W = -1
  largestX_B = -1
  width, height = img.size

  img_pix_sum = 0

  for y in range(height):
      largestX_W = -1
      largestX_B = -1
      first_W = False
      first_B = False
      for x in range(width): #read each pixel group
          if(x > largestX_W and pixels[x, y] == 255):
              img_pix_sum += 1
              xsize = 1
              seenBlack = False
              while(x+xsize < width and seenBlack == False):
                  if(pixels[x+xsize, y] == 255):
                      img_pix_sum += 1
                      xsize += 1
                      largestX_W = x+xsize
                  else:
                      seenBlack = True
              if(not first_W):
                first_W = True
                stringout_w += f"{x},{xsize}"
              else:
                stringout_w += f",{x},{xsize}"
          elif(x > largestX_B and pixels[x, y] == 0):
              xsize = 1
              seenWhite = False
              while(x+xsize < width and seenWhite == False):
                  if(pixels[x+xsize, y] == 0):
                      xsize += 1
                      largestX_B = x+xsize
                  else:
                      seenWhite = True
              if(not first_B):
                first_B = True
                stringout_b += f"{x},{xsize}"
              else:
                stringout_b += f",{x},{xsize}"
      if(y != height-1):
        stringout_b += 'n'
        stringout_w += 'n'

  if(img_pix_sum / (width * height) >= 0.5):
    stringout += stringout_b
  else:
    stringout += stringout_w
  stringout += 'n'
with open("animation.txt", "w", encoding="utf-8") as file:
    file.write(stringout)
```
