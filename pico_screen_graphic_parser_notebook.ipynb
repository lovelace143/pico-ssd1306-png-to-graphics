```python
from PIL import Image

#replace handOn with your image's filename, ofc
filename = 'handOn'
img = Image.open(filename+'.png')
img = img.convert('RGBA')
myWidth = img.width
pixels = img.load()
stringout = ''
xsize = 1
largestX = -1
width, height = img.size

for y in range(height):
    largestX = -1
    first = False
    for x in range(width): #read each pixel group
        if(x > largestX and pixels[x, y] == (255, 255, 255, 255)):
            r, g, b, a = pixels[x, y]
            xsize = 1
            seenBlack = False
            while(x+xsize < width and seenBlack == False):
                if(pixels[x+xsize, y] == (255, 255, 255, 255)):
                    xsize += 1
                    largestX = x+xsize
                else:
                    seenBlack = True
            if(not first):
              first = True
              stringout += f"{x},{xsize}"
            else:
              stringout += f",{x},{xsize}"
    if(y != height-1):
      stringout += '\n'

with open(filename+".txt", "w", encoding="utf-8") as file:
    file.write(stringout)
```

```python
#code example of how these should be read in the PICO under micropython:
print("display = ssd1306.SSD1306_SPI(...)")
lines = stringout.split("\n")
yPos = 0
for line in range(len(lines)):
  lineArr = lines[line].split(",")
  for i in range(int(len(lineArr)/2)): #automatically skips empty lines which get rounded to 0
    xPos = int(lineArr[(i*2)])
    xSize = int(lineArr[(i*2)+1])
    print(f"display.fill_rect({xPos}, {yPos}, {xSize}, 1, 1)")
  yPos += 1
```
